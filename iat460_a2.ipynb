## L-System Implementation

```python
# Season Dictionaries
spring = {
    "leaves": ["#C0DE7B", "#79B425", "#1D6517"],
    "trunk_color": "saddlebrown",
    "trunk_width": 2,
    "bg_color": "#78f6ff",
    "leaf_size": 20
}

summer = {
    "leaves": ["#658406", "#4b6702", "#3c5003"],
    "trunk_color": "saddlebrown",
    "trunk_width": 2,
    "bg_color": "#2e7eff",
    "leaf_size": 20
}

fall = {
    "leaves": ["#4C1B16", "#A11913", "#DE451D", "#EBC63F"],
    "trunk_color": "saddlebrown",
    "trunk_width": 2,
    "bg_color": "#217eff",
    "leaf_size": 15
}

winter = {
    "leaves": ["#FFFFFF"],
    "trunk_color": "saddlebrown",
    "trunk_width": 1,
    "bg_color": "#efefef",
    "leaf_size": 0
}

cherry_blossom = {
    "leaves": ["#9A3050", "#F5C0CC", "#C84870", "#E890A8", "#FAEAEE"],
    "trunk_color": "thistle",
    "trunk_width": 4,
    "bg_color": "#5A86CB",
    "leaf_size": 25
}

# Based on the week 3 lab code: https://colab.research.google.com/github/IAT-ComputationalCreativity-Spring2026/Lab-3/blob/main/l-systems.ipynb#scrollTo=pOFesYNwkSkC
def create_l_system(iterations, axiom, rules):
    """Generate L-System instructions based on axiom and rules."""
    result = axiom
    for _ in range(iterations):
        new_string = ""
        for char in result:
            new_string += rules.get(char, char)
        result = new_string
    return result

def draw_l_system(instructions, angle, distance):
    """Draw the L-System using turtle graphics.

    Parameters:
    - instructions: string of L-System commands
    - angle: turning angle in degrees
    - distance: forward movement distance
    """
    stack = []
    for cmd in instructions:
        if cmd == 'R':
            t.forward(distance)
        elif cmd == 'F':  # Move forward and draw
            t.forward(distance)
        elif cmd == 'f':  # Move forward without drawing
            t.penup()
            t.forward(distance)
            t.pendown()
        elif cmd == '+':  # Turn right
            t.right(angle)
        elif cmd == '-':  # Turn left
            t.left(angle)
        elif cmd == '[':  # Save current state
            stack.append((t.position(), t.heading()))
        elif cmd == ']':  # Restore previous state
            # choose a random leaf colour from the corresponding season palette
            t.pencolor(random.choice(season["leaves"]))
            t.begin_fill()
            t.dot(season["leaf_size"])
            t.end_fill()

            position, heading = stack.pop()
            t.penup()
            t.goto(position)
            t.setheading(heading)
            t.pendown()
            t.pencolor(season["trunk_color"])

def setup_turtle():
        t.initializeTurtle(mode="svg")
        t.hideturtle()
        t.speed(13)  # Fastest speed
        t.penup()
        t.bgcolor(season["bg_color"])
        t.goto(t.window_width() // 2, t.window_height() - 50)  # Start position
        t.pendown()
```

##Examples: The 4 Seasons of Trees

By changing the parameters of the ruleset, you can create a tree for each season (Spring, Summer, Fall, Winter).

Spring: The leaves are small and light green, and only a few branches have grown in.

Summer: The leaves are a deep green,they appear denser, and the tree has more branches.

Fall: The leaves are various colours of reds, oranges, and yellows. Also there is less of them as they have begun to fall off for the winter.

Winter: There are no leaves, and the tree is thinner than it normally is.

(Bonus) Cherry Blossom Tree: There are 5 different types of pick cherry blossoms on the tree

### Example 1: Spring Tree

```python
# Set season here using string: spring, summer, fall, winter, cherry_blossom
season = spring

# Your fractal tree parameters
tree_axiom = "F"  # Start with a single trunk

# Rule 1
# tree_rules = {"F": "F[+F]F[+][+]F[-][-F]F[-F][-]F"}

# Rule 2
tree_rules = {"F": "F[+FFF]F[-FF][+FF]F[-F][+FF]FF[-FFF]F"}

tree_iterations = 2
tree_angle = 30

# Generate and draw
setup_turtle()
t.pensize(season['trunk_width'])
t.pencolor(season['trunk_color'])
tree_instructions = create_l_system(tree_iterations, tree_axiom, tree_rules)
draw_l_system(tree_instructions, tree_angle, 15)

#setting colour for the final leaf on top
t.pencolor(random.choice(season["leaves"]))

t.begin_fill()
t.dot(season["leaf_size"])
t.end_fill()
```
